### **Final Step, Now that you understand every preprocessing technique individually**

1) build a single preprocessing pipeline.

2) Finally, combine them using ColumnTransformer, refer to    [documentation](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html) and transform the complete dataset.

```python
# Step1

import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    StandardScaler,
    RobustScaler,
    MinMaxScaler,
    OrdinalEncoder,
    OneHotEncoder
)

# Numerical columns
standard_features = ['Age']
robust_features = ['Salary']
minmax_features = ['Experience']

# Ordinal column
ordinal_features = ['Education']

# One-Hot Encoding column
onehot_features = ['City']

# Binary column
gender_features = ['Gender']

standard_transformer = StandardScaler()

robust_transformer = RobustScaler()

minmax_transformer = MinMaxScaler()

ordinal_transformer = OrdinalEncoder(
    categories=[['High School', 'Bachelor', 'Master', 'PhD']]
)

onehot_transformer = OneHotEncoder(
    drop='first',
    sparse_output=False
)

# Binary Encoding using OrdinalEncoder
gender_transformer = OrdinalEncoder(
    categories=[['Male', 'Female']]
)

preprocessor = ColumnTransformer(
    transformers=[
        ('age_scaling', standard_transformer, standard_features),
        ('salary_scaling', robust_transformer, robust_features),
        ('experience_scaling', minmax_transformer, minmax_features),
        ('education_encoding', ordinal_transformer, ordinal_features),
        ('gender_encoding', gender_transformer, gender_features),
        ('city_encoding', onehot_transformer, onehot_features)
    ]
)

processed_data = preprocessor.fit_transform(df)

city_columns = preprocessor.named_transformers_[
    'city_encoding'
].get_feature_names_out(['City'])

final_columns = [
    'Age',
    'Salary',
    'Experience',
    'Education',
    'Gender'
] + list(city_columns)

processed_df = pd.DataFrame(
    processed_data,
    columns=final_columns
)

print(processed_df)
```
