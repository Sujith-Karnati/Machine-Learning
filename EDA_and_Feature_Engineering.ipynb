# Text Preprocessing

```python
pip install contractions
```

```python
import contractions
df["Text"] = df["Text"].apply(lambda x: contractions.fix(str(x)))  # convert the 'don't' --> 'do not'
df['Summary'] = df["Summary"].apply(lambda x: contractions.fix(str(x)))
```

```python
import pandas as pd
import numpy as np
import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download required NLTK resources
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize stopwords and lemmatizer
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

# Text preprocessing function
def preprocess(text):

    # Convert to string
    text = str(text)

    # Lowercase
    text = text.lower()

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenization
    tokens = text.split()

    # Remove stopwords
    tokens = [word for word in tokens if word not in stop_words]

    # Lemmatization
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    # Join back into a sentence
    return " ".join(tokens)

# Apply preprocessing
df['Summary'] = df['Summary'].apply(preprocess)
df['Text'] = df['Text'].apply(preprocess)

# Display processed data
print(df[['Summary', 'Text']].head())
```

## Exploratory Data Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Distribution of Ratings


```python
dist_ratings = df['Score'].groupby(df['Score']).count()
plt.figure(figsize=(10,6))
sns.barplot(x=dist_ratings.index, y=dist_ratings.values)
plt.title('Distribution of Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

## Rating Percentage Distribution


```python
rating_percent = (df['Score'].value_counts(normalize=True).sort_index() * 100).round(2)
plt.figure(figsize = (10,6))
sns.barplot(x = rating_percent.index, y = rating_percent.values)
plt.title('Rating Percentage Distribution')
plt.xlabel('Rating')
plt.ylabel('Percentage')
plt.show()
```

## Most Frequent Words in Positive Reviews (4–5 Ratings)


```python
positive_reviews = df[df['Score'] >= 4]['Text']
positive_words = ' '.join(positive_reviews)



from collections import Counter

words = positive_words.split()

word_counts = Counter(words)

top_words = word_counts.most_common(20)

print(top_words)


top_positive = pd.DataFrame(
    top_words,
    columns=['Word', 'Frequency']
)

print(top_positive)

import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.bar(top_positive['Word'],
        top_positive['Frequency'],
        color='green',
        edgecolor='black')

plt.title("Top 20 Most Frequent Words in Positive Reviews")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation=45)

plt.show()
```

## Most Frequent Words in Negative Reviews (1-2 Ratings)


```python
negative_reviews = df[df['Score'] < 3]['Text']
negative_words = ' '.join(negative_reviews)
words = negative_words.split()

word_counts = Counter(words)

top_words = word_counts.most_common(20)
top_words

top_negative = pd.DataFrame(
    top_words,
    columns = ['word','frequency']
)

print(top_negative)

plt.figure(figsize = (12,6))
plt.bar(top_negative['word'],top_negative['frequency'],color = 'red',edgecolor = 'black')
plt.title("Top 20 Most Frequent Words in Negative Reviews")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation = 45)
plt.show()
```

## Distribution of Helpfulness Ratio


```python
df['Helpfulness_Ratio'] = np.where(
    df['HelpfulnessDenominator'] == 0,
    0,
    df['HelpfulnessNumerator'] / df['HelpfulnessDenominator']
)

df[['HelpfulnessNumerator','HelpfulnessDenominator','Helpfulness_Ratio']].head()

plt.figure(figsize=(10,6))

plt.hist(
    df['Helpfulness_Ratio'],
    bins=30,
    color='skyblue',
    edgecolor='black'
)

plt.title("Distribution of Helpfulness Ratio")
plt.xlabel("Helpfulness Ratio")
plt.ylabel("Number of Reviews")

plt.show()
```

## Relationship between Rating and Helpfulness


```python
rating_helpfulness = df.groupby('Score')['Helpfulness_Ratio'].mean().reset_index()

print(rating_helpfulness)

plt.figure(figsize=(8,5))

sns.barplot(
    data=rating_helpfulness,
    x='Score',
    y='Helpfulness_Ratio',
    palette='viridis'
)

plt.title("Average Helpfulness Ratio by Rating")
plt.xlabel("Rating")
plt.ylabel("Average Helpfulness Ratio")
plt.show()
```

## Review Length Distribution


```python
# Review length (Number of words)
df['Review_Length'] = df['Text'].apply(lambda x: len(str(x).split()))

# Display first few values
df[['Text', 'Review_Length']].head()

plt.figure(figsize=(10,6))

sns.histplot(
    df['Review_Length'],
    bins=50,
    kde=True,
    color='skyblue'
)

plt.title("Distribution of Review Length")
plt.xlabel("Review Length (Number of Words)")
plt.ylabel("Frequency")

plt.show()
```

```python
avg_review_length = df.groupby('Score')['Review_Length'].mean().reset_index()

print(avg_review_length)


plt.figure(figsize=(8,5))

sns.barplot(
    data=avg_review_length,
    x='Score',
    y='Review_Length',
    palette='viridis'
)

plt.title("Average Review Length by Rating")
plt.xlabel("Rating")
plt.ylabel("Average Review Length (Words)")

# Display values on top of bars
for index, value in enumerate(avg_review_length['Review_Length']):
    plt.text(index, value + 1, f"{value:.1f}", ha='center')

plt.show()
```
