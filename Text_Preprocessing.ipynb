```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()

words = ['playing', 'played','plays','studies','historical','running']

for word in words:
  print(word, '--->', stemmer.stem(word))
```

```python
import nltk
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()

words = ['mice','better','studies','running']

for word in words:
  print(word, '--->', lemmatizer.lemmatize(word))
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize('running', pos=wordnet.VERB))
print(lemmatizer.lemmatize('better', pos=wordnet.ADJ))
```

### OneHotEncoding

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
```

```python
words = pd.DataFrame({
    'word' : ['I','Love','Machine','Learning']
})

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(words)

print(encoded)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = [
    'I love NLP',
    'I love machine learning',
    'machine learning is amazing'
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    'I love NLP',
    'I love machine learning',
    'machine learning is amazing'
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```
