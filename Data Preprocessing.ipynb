## Data Preprocessing

```python
import pandas as pd
```

### DataFrame creation

```python
df = pd.DataFrame({
    'Age':[25,23,26,28,24],
    'Salary':[45000,50000,67000,56000,90000],
    'Gender':['Male','Female','Female','Male','Female'],
    'City':['Delhi','Mumbai',"Chennai",'Mumbai','Hyderabad'],
    'Education':['UG','PG','PHD',"PG",'UG']
})
```

### Data understanding


```python
df.info()
```

### checking the data

```python
df
```

### importing the preprocessing library

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

## Numerical data pre processing

### Standardizing the numerical columns

```python
x = df[['Age','Salary']]
scaler = StandardScaler()
scale_data = scaler.fit_transform(x)
scale_data
```

### Normalizing the numerical values

```python
scaler = MinMaxScaler()
scale_data = scaler.fit_transform(x)
scale_data
```

## Categorical data transformation
### One HotEncoding

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder(sparse_output=False)
encoded=encoder.fit_transform(df[['Gender','City']])
encoded
```

### ordinal encoading

```python
from sklearn.preprocessing import OrdinalEncoder
encoder=OrdinalEncoder(categories=[['UG','PG','PHD']])
encoded=encoder.fit_transform(df[['Education']])
encoded
```

### lable encoding

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
encoded=encoder.fit_transform(df['Gender'])
encoded
```

```python
df['Gender_lable1'] = encoder.fit_transform(df['Gender'])
```
